```python
numerical_features=[
    "Mileage",
    "Vehicle_Age",
    "Engine_Size",
    "Odometer_Reading",
    "Insurance_Premium",
    "Service_History",
    "Accident_History",
    "Fuel_Efficiency",
    "Days_Since_Last_Service",
    "Days_Until_Warranty_Expiry"
]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder,StandardScaler
```

```python
numeric_transformer=Pipeline(steps=[
    ("imputer",SimpleImputer(strategy="median")),
    ("scaler",StandardScaler())
])
categorical_transformer=Pipeline(steps=[
    ("imputer",SimpleImputer(strategy="most_frequent")),
    ("onehot",OneHotEncoder(handle_unknown="ignore"))
])
```

```python
preprocessor=ColumnTransformer(
    transformers=[
        ("num",numeric_transformer,numerical_features),
        ("cat",categorical_transformer,categorical_features)
    ]
)
```

```python
#Test The Processing
```

```python
X_train_processed=preprocessor.fit_transform(X_train)
X_test_processed=preprocessor.transform(X_test)
```

```python
print("Original training shape:",X_train.shape)
print("Processed training shape:",X_train_processed.shape)
print("Original testing shape:",X_test.shape)
print("Processed testing shape:",X_test_processed.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,StandardScaler
date_columns=[
    "Last_Service_Date",
    "Warranty_Expiry_Date"
]
X_train_clean=X_train.drop(columns=date_columns,errors="ignore")
X_test_clean=X_test.drop(columns=date_columns,errors="ignore")
categorical_features=X_train_clean.select_dtypes(
    include=["object","category"]
).columns.tolist()
numerical_features=X_train_clean.select_dtypes(
    include=["int64","float64"]
).columns.tolist()
print("Categorical Features:")
print(categorical_features)
print("\nNumerical Features:")
print(numerical_features)
preprocessor=ColumnTransformer(
    transformers=[
        ("num",StandardScaler(),numerical_features),
        ("cat",OneHotEncoder(handle_unknown="ignore"),categorical_features)
    ]
)
X_train_processed=preprocessor.fit_transform(X_train_clean)
X_test_processed=preprocessor.transform(X_test_clean)
print("\nPreproceesing complete")
print("\nProcessed Training Data:")
print("X_train_processed shape:",X_train_processed.shape)
print("\nProcessed Testing Data:")
print("X_test_processed shape:",X_test_processed.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import(accuracy_score,confusion_matrix,
                            classification_report,
                            balanced_accuracy_score,
                            roc_auc_score,
                            precision_score,recall_score,f1_score)
balanced_lr=LogisticRegression(
    max_iter=1000,class_weight="balanced",random_state=42)
balanced_lr.fit(X_train_processed,y_train)
y_pred=balanced_lr.predict(X_test_processed)
y_prob=balanced_lr.predict_proba(X_test_processed)[:,1]
accuracy=accuracy_score(y_test,y_pred)
balanced_acc=balanced_accuracy_score(y_test,y_pred)
roc_auc=roc_auc_score(y_test,y_prob)
precision=precision_score(y_test,y_pred)
recall=recall_score(y_test,y_pred)
f1=f1_score(y_test,y_pred)
print("======Logistic Regression Results======")
print("Accuracy:",accuracy)
print("Balanced Accuracy:",balanced_acc)
print("ROC-AUC:",roc_auc)
print("Precision:",precision)
print("Recall:",recall)
print("F1-Score:",f1)
print("\nConfusion Matrix:")
print(confusion_matrix(y_test,y_pred))
print("\nClassification Report:")
print(classification_report(y_test,y_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier
decision_tree=DecisionTreeClassifier(class_weight="balanced",random_state=42,max_depth=10)
decision_tree.fit(X_train_processed,y_train)
dt_pred=decision_tree.predict(X_test_processed)
dt_prob=decision_tree.predict_proba(X_test_processed)[:,1]
dt_accuracy=accuracy_score(y_test,dt_pred)
dt_balanced_accuracy=balanced_accuracy_score(y_test,dt_pred)
dt_roc_auc=roc_auc_score(y_test,dt_prob)
dt_precision=precision_score(y_test,dt_pred)
dt_recall=recall_score(y_test,dt_pred)
dt_f1=f1_score(y_test,dt_pred)
print("======Balanced Decision Tree =======")
print(f"\nAccuracy:{dt_accuracy:.4f}")
print(f"Balanced Accuracy:{dt_balanced_accuracy:.4f}")
print(f"ROC-AUC:{dt_roc_auc:.4f}")
print(f"Precision:{dt_precision:.4f}")
print(f"Recall:{dt_recall:.4f}")
print(f"F1-Score:{dt_f1:.4f}")
print(" Confusion Matrix:")
print(confusion_matrix(y_test,dt_pred))
print("\nClassification Report:")
print(classification_report(y_test,dt_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
random_forest_clean=RandomForestClassifier(class_weight="balanced",random_state=42,n_estimators=200,n_jobs=-1)
random_forest_clean.fit(X_train_processed,y_train)
rf_pred=random_forest_clean.predict(X_test_processed)
rf_prob=random_forest_clean.predict_proba(X_test_processed)[:,1]
rf_accuracy=accuracy_score(y_test,rf_pred)
rf_balanced_accuracy=balanced_accuracy_score(y_test,rf_pred)
rf_roc_auc=roc_auc_score(y_test,rf_prob)
rf_precision=precision_score(y_test,rf_pred)
rf_recall=recall_score(y_test,rf_pred)
rf_f1=f1_score(y_test,rf_pred)
print("====Balanced Random Forest====")
print(f"\nAccuracy:{rf_accuracy:.4f}")
print(f"Balanced Accuracy:{rf_balanced_accuracy:.4f}")
print(f"ROC-AUC:{rf_roc_auc:.4f}")
print(f"Precision:{rf_precision:.4f}")
print(f"Recall:{rf_recall:.4f}")
print(f"F1-Score:{rf_f1:.4f}")
print(" Confusion Matrix:")
print(confusion_matrix(y_test,rf_pred))
print("\nClassification Report:")
print(classification_report(y_test,rf_pred))
```

```python
import pandas as pd
import numpy as np
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)
from IPython.display import display as show_table

# Calculate predictions using your existing trained model
test_probabilities = balanced_lr_clean.predict_proba(
    X_test_clean_processed
)[:, 1]

test_predictions = (test_probabilities >= 0.35).astype(int)

# Calculate confusion matrix
cm_calculated = confusion_matrix(
    y_test, test_predictions, labels=[0, 1]
)

tn, fp, fn, tp = cm_calculated.ravel()

# Calculate evaluation metrics
metrics = [
    ("Accuracy", accuracy_score(y_test, test_predictions)),
    ("Balanced accuracy", balanced_accuracy_score(y_test, test_predictions)),
    ("Maintenance precision", precision_score(
        y_test, test_predictions, pos_label=1, zero_division=0
    )),
    ("Maintenance recall", recall_score(
        y_test, test_predictions, pos_label=1, zero_division=0
    )),
    ("Maintenance F1-score", f1_score(
        y_test, test_predictions, pos_label=1, zero_division=0
    )),
    ("Specificity (class 0 recall)", tn / (tn + fp)),
    ("False positives", fp),
    ("False negatives", fn)
]

test_metrics = pd.DataFrame(metrics, columns=["Metric", "Value"])

# Format percentages and counts correctly
test_metrics["Value"] = test_metrics.apply(
    lambda row: (
        f"{row['Value']:.2%}"
        if row["Metric"] not in ["False positives", "False negatives"]
        else f"{int(row['Value']):,}"
    ),
    axis=1
)

show_table(test_metrics)

# Save the table for your research paper
test_metrics.to_csv("Table3_Test_Metrics.csv", index=False)

print("Table saved as Table3_Test_Metrics.csv")
```
